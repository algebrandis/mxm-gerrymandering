```python
dem_data = {}
```

```python
for x in dem_list:
    dem_data[x] = []
dem_data
```

```python
for n in range(6, 287):
    listn = list(dem_election.iloc[n])
    for x in range(0, len(listn)):
        dem_data[dem_list[x]].append(listn[x])
```

```python
dem_df = pd.DataFrame(dem_data)
dem_df
```

```python
rep_list = list(rep_election.iloc[5])
rep_data = {}
for x in rep_list:
    rep_data[x] = []
rep_data
for n in range(6, 287):
    listn = list(rep_election.iloc[n])
    for x in range(0, len(listn)):
        rep_data[rep_list[x]].append(listn[x])
rep_df = pd.DataFrame(rep_data)
rep_df
```

```python
election_df = pd.merge(dem_df, rep_df, on = "Precincts", how = "inner")
election_df
```
